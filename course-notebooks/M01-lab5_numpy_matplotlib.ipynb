## Part 1: Matplotlib Basics

Matplotlib is Python's most widely used plotting library. You'll use it constantly — for visualizing data, debugging models, and presenting results.

The module we use is `matplotlib.pyplot`, which we import as `plt`. The basic pattern is simple:

1. Create some data (usually NumPy arrays)
2. Call a plotting function (`plt.plot()`, `plt.scatter()`, etc.)
3. Add labels, titles, and other formatting
4. Call `plt.show()` to display it (Colab often does this automatically, but it's good practice)

Let's start with line plots.

### Line Plots with `plt.plot()`

`plt.plot(x, y)` draws a line connecting the (x, y) points you give it. Both `x` and `y` should be arrays (or lists) of the same length. Run this:

````python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 1, 5, 3])

plt.plot(x, y)
plt.show()
````

That's the minimum. But a plot without labels is useless — nobody knows what they're looking at. Let's add labels and a title:

````python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 1, 5, 3])

plt.plot(x, y)
plt.xlabel("Time (seconds)")
plt.ylabel("Value")
plt.title("My First Plot")
plt.show()
````

### Multiple Lines and Legends

You can call `plt.plot()` multiple times before `plt.show()` to put several lines on the same graph. Use the `label` parameter and `plt.legend()` so viewers can tell them apart:

````python
x = np.array([1, 2, 3, 4, 5])
y1 = np.array([2, 4, 1, 5, 3])
y2 = np.array([1, 3, 5, 2, 4])

plt.plot(x, y1, label="Sensor A")
plt.plot(x, y2, label="Sensor B")
plt.xlabel("Time (seconds)")
plt.ylabel("Reading")
plt.title("Two Sensors")
plt.legend()
plt.show()
````

### Controlling Figure Size

If a plot looks cramped, use `plt.figure(figsize=(width, height))` *before* your plot calls. The units are inches:

````python
plt.figure(figsize=(10, 5))

x = np.linspace(0, 10, 100)  # 100 evenly spaced points from 0 to 10
plt.plot(x, np.sin(x))
plt.title("A wider plot")
plt.show()
````

## Part 2: Other Plot Types

Line plots are great for continuous data, but sometimes you need different visualizations. Here are three more plot types you'll use all the time.

### Bar Charts — `plt.bar()`

Use bar charts when you're **comparing categories** — things that aren't on a continuous scale. The first argument is the category names, the second is the values:

````python
fruits = ["Apples", "Bananas", "Cherries", "Dates"]
sales = [45, 32, 58, 17]

plt.bar(fruits, sales, color="steelblue")
plt.ylabel("Units Sold")
plt.title("Fruit Sales")
plt.show()
````

### Histograms — `plt.hist()`

Use histograms to see the **distribution** of a dataset — how often values fall into different ranges. You give it the raw data and a number of `bins` (buckets). It counts for you:

````python
# 500 random test scores centered around 75
scores = np.random.normal(loc=75, scale=10, size=500)

plt.hist(scores, bins=20, color="coral", edgecolor="black")
plt.xlabel("Score")
plt.ylabel("Count")
plt.title("Test Score Distribution")
plt.show()
````

That bell shape is a **normal distribution** — you'll see it everywhere in data science and ML.

### Scatter Plots — `plt.scatter()`

Use scatter plots to see the **relationship between two variables**. Each point is one observation with an x and y value:

````python
# Height vs shoe size (made-up data with some noise)
height = np.random.normal(170, 10, 50)
shoe_size = height * 0.1 + np.random.normal(0, 1, 50)

plt.scatter(height, shoe_size, color="teal", alpha=0.7)
plt.xlabel("Height (cm)")
plt.ylabel("Shoe Size")
plt.title("Height vs Shoe Size")
plt.show()
````

The upward trend shows these variables are **correlated** — taller people tend to have bigger feet. That's the kind of pattern scatter plots reveal.
